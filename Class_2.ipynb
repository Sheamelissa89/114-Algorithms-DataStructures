```python
class Calculator:
    def __init__(self, number1, number2):
        self.number1 = number1
        self.number2 = number2

    def add(self):
        return self.number1 + self.number2    

    def subtract(self):
        return self.number1 - self.number2

    def multiply(self):
        return self.number1 * self.number2

    def divide(self):
        if self.number2 == 0:
            return ("Cannot divide by zero.")
        return self.number1 / self.number2

    def __str__(self):
        return f"Calculator({self.number1}, {self.number2})"
```

```python
c1 = Calculator(5, 2)

print(c1)
print(f"Addition: {c1.add()}")
print(f"Subtraction: {c1.subtract()}")
print(f"Multiplication: {c1.multiply()}")
print(f"Division: {c1.divide()}")
```

```python
c2 = Calculator(10, 0)

print(c2)
print(f"Division: {c2.divide()}")
```
