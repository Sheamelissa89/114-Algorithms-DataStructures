```python
def count_even_numbers(collection):
    total = 0

    for number in collection:
        if number % 2 == 0:
            total += 1

    return total        
```

```python
numbers = [3, 5, 27, 20, 6, 3.5, 60]

total = count_even_numbers(numbers)

print(f"total = {total}")
```

```python
numbers = ["1", "2"]

total = count_even_numbers(numbers)

print(f"total = {total}")
```
