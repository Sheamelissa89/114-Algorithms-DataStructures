```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

    def __str__(self):
        return f"Node({self.data})"

class CustomQueue:
    def __init__(self):
        self.front = None
        self.rear = None

    def is_empty(self):
        return self.front is None

    def enqueue(self, data):
        new_node = Node(data)

        if self.is_empty():
            self.front = new_node
            self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node

    def dequeue(self):
        if self.is_empty():
            raise IndexError("Queue is Empty")

        data = self.front.data
        self.front = self.front.next

        if self.front is None:
            self.rear = None

        return data

    # Final Report methods

    def __str__ (self):
        elements = []
        current = self.front

        while current is not None:
            elements.append(str(current.data))
            current = current.next

        return f"CustomQueue -> [{', '.join(elements)}]"

    def peek(self):
        if self.is_empty():
            raise IndexError("Queue is Empty")

        return self.front.data

    def size(self):
        count = 0
        current = self.front

        while current is not None:
            count += 1
            current = current.next

        return count     
```

```python
queue = CustomQueue()

queue.enqueue("A")
queue.enqueue("B")
queue.enqueue("C")

print(queue) 
print("Front value:", queue.peek())
print("Queue size:", queue.size())
```

```python
print("Removed value:", queue.dequeue())
print("Queue after dequeue:", queue)
print("New front value:", queue.peek())
print("New queue size:", queue.size())
```
